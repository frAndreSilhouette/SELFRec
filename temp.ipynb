```python
import re

modified_models = ['NCL', 'SGL', 'SimGCL', 'XSimGCL']
campus_ids = [10, 15, 34, 102, 143]

for model in modified_models:
    yaml_path = f'./conf/{model}.yaml'
    with open(yaml_path, 'r', encoding='utf-8') as f:
        yaml_content = f.read()

    for campus_id in campus_ids:
        # 替换 campus_*
        new_yaml_str = re.sub(r"campus_(\*|\d+)", f"campus_{campus_id}", yaml_content)

        # 新文件路径
        new_yaml_path = f'./conf/{model}_{campus_id}.yaml'
        with open(new_yaml_path, 'w', encoding='utf-8') as f:
            f.write(new_yaml_str)

        print(f"Generated: {new_yaml_path}")
```
