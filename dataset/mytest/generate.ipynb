```python
import random
import string

# 配置
num_users = 1000      # 用户数
num_items = 2000     # 商品数
num_records = 1000000  # 总交互数
train_ratio = 0.8    # 训练集比例

train_file = 'train.txt'
test_file = 'test.txt'

# 生成字符串ID
def random_id(prefix, n=5):
    return prefix + ''.join(random.choices(string.ascii_letters + string.digits, k=n))

user_ids = [random_id('user') for _ in range(num_users)]
item_ids = [random_id('item') for _ in range(num_items)]

data = []

for _ in range(num_records):
    user_id = random.choice(user_ids)
    item_id = random.choice(item_ids)

    delta_time = random.randint(1, 30)  # 间隔天数
    scale = round(random.uniform(1.0, 10.0), 3)   # Weibull scale
    shape = round(random.uniform(0.5, 2.0), 3)    # Weibull shape
    weight = 1  # 固定权重

    data.append([user_id, item_id, delta_time, scale, shape, weight])

# 打乱顺序
random.shuffle(data)

# 切分训练集和测试集
split_idx = int(len(data) * train_ratio)
train_data = data[:split_idx]
test_data = data[split_idx:]

# 写入文件
with open(train_file, 'w') as f:
    for row in train_data:
        f.write(" ".join(map(str, row)) + "\n")

with open(test_file, 'w') as f:
    for row in test_data:
        f.write(" ".join(map(str, row)) + "\n")

print(f"生成完成: {train_file} ({len(train_data)}行), {test_file} ({len(test_data)}行)")
```
