```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 读取数据
df = pd.read_csv('/mnt/workspace/推荐算法forRecSys/weibull_params.csv')

sns.set(style="whitegrid")

def plot_with_stats(column, color, ax):
    # 绘制直方图 + KDE
    sns.histplot(df[column], kde=True, bins=30, color=color, ax=ax)
    ax.set_title(f'{column.capitalize()} Distribution')
    ax.set_xlabel(column.capitalize())
    ax.set_ylabel('Frequency')
    
    # 计算统计指标
    stats = df[column].describe()[['25%', '50%', '75%', 'mean']]
    
    # 将统计指标添加到图中
    stats_text = '\n'.join([f'{k}: {v:.3f}' for k, v in stats.items()])
    ax.text(0.95, 0.95, stats_text, transform=ax.transAxes,
            verticalalignment='top', horizontalalignment='right',
            bbox=dict(boxstyle='round,pad=0.5', facecolor='white', alpha=0.8))

# 创建图形
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

plot_with_stats('scale', 'skyblue', axes[0])
plot_with_stats('shape', 'salmon', axes[1])

plt.tight_layout()
plt.show()
```
